```python
scores = [] # ใช้ชื่อตัวแปรให้สื่อความหมายมากขึ้น
num_scores = 5

for i in range(num_scores):
    # รับค่าคะแนนทีละตัวใน Loop
    try:
        score = int(input(f"กรอกคะแนนค่าที่ {i + 1}: "))
        scores.append(score)
    except ValueError:
        print("⚠️ กรุณากรอกเป็นจำนวนเต็มเท่านั้น!")

print(f"\nList ที่เก็บคะแนน: {scores}")
print("--- แสดงผลคะแนนแต่ละค่า ---")

# Loop จะทำงานตามจำนวนคะแนนที่รับมา
for g in scores:
    print("คะแนนคือ:", g)
```
